**Cenário de negócio:** a empresa libera uma oferta de fidelização somente se o cliente estiver com o pagamento em dia **e** tiver alto engajamento (porta lógica AND).

# CP1 – Códigos
## 1. Dados do problema

```python
X = [
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
]

y = [0, 0, 0, 1]

print("Registros (pagamento em dia, alto engajamento -> liberar oferta):")
for i in range(len(X)):
    print(f"Registro {i+1}: entrada = {X[i]} | esperado = {y[i]}")
```

## 2. Função de ativação (degrau)

```python
def degrau(valor):
    if valor >= 0:
        return 1
    else:
        return 0

# teste rápido
print(degrau(-0.5), degrau(0), degrau(0.3))
```

## 3. Perceptron sem viés
Pesos iniciais `[0.0, 0.0]`, taxa de aprendizado `0.1`, 10 épocas.

```python
pesos = [0.0, 0.0]
taxa = 0.1
epocas = 10

historico_sem_vies = []  # erros por época

for epoca in range(1, epocas + 1):
    erros_epoca = 0
    print(f"\n===== ÉPOCA {epoca} =====")
    for i in range(len(X)):
        entrada = X[i]
        esperado = y[i]

        soma = entrada[0] * pesos[0] + entrada[1] * pesos[1]
        previsto = degrau(soma)
        erro = esperado - previsto

        pesos_antes = pesos.copy()

        pesos[0] = pesos[0] + taxa * erro * entrada[0]
        pesos[1] = pesos[1] + taxa * erro * entrada[1]

        if erro != 0:
            erros_epoca += 1

        print(f"Época {epoca} | Entrada: {entrada} | Esperado: {esperado} | Previsto: {previsto} | "
              f"Erro: {erro} | Pesos antes: {[round(p,2) for p in pesos_antes]} | "
              f"Pesos depois: {[round(p,2) for p in pesos]}")

    historico_sem_vies.append(erros_epoca)
    print(f"Quantidade de erros na época {epoca}: {erros_epoca}")

print("\nErros por época (sem viés):", historico_sem_vies)
print("Pesos finais (sem viés):", [round(p, 2) for p in pesos])
```

## 4. Perceptron com viés
O treinamento é encerrado quando uma época termina com zero erros.

```python
pesos = [0.0, 0.0]
bias = 0.0
taxa = 0.1
epocas = 10

historico_com_vies = []
epoca_convergencia = None

for epoca in range(1, epocas + 1):
    erros_epoca = 0
    print(f"\n===== ÉPOCA {epoca} =====")
    for i in range(len(X)):
        entrada = X[i]
        esperado = y[i]

        soma = entrada[0] * pesos[0] + entrada[1] * pesos[1] + bias
        previsto = degrau(soma)
        erro = esperado - previsto

        pesos_antes = pesos.copy()
        bias_antes = bias

        pesos[0] = pesos[0] + taxa * erro * entrada[0]
        pesos[1] = pesos[1] + taxa * erro * entrada[1]
        bias = bias + taxa * erro

        if erro != 0:
            erros_epoca += 1

        print(f"Época {epoca} | Entrada: {entrada} | Esperado: {esperado} | Previsto: {previsto} | "
              f"Erro: {erro} | Pesos antes: {[round(p,2) for p in pesos_antes]} | "
              f"Pesos depois: {[round(p,2) for p in pesos]} | "
              f"Bias antes: {round(bias_antes,2)} | Bias depois: {round(bias,2)}")

    historico_com_vies.append(erros_epoca)
    print(f"Quantidade de erros na época {epoca}: {erros_epoca}")

    if erros_epoca == 0:
        epoca_convergencia = epoca
        print(f"\n>>> Convergiu na época {epoca} (zero erros).")
        break

print("\nErros por época (com viés):", historico_com_vies)
print("Pesos finais:", [round(p, 2) for p in pesos])
print("Bias final:", round(bias, 2))
```

```python
# Verificação final do modelo com viés
print("Verificação final:")
for i in range(len(X)):
    soma = X[i][0]*pesos[0] + X[i][1]*pesos[1] + bias
    print(f"Entrada {X[i]} -> soma = {soma:.2f} -> previsto = {degrau(soma)} (esperado {y[i]})")
```

# CP2 – Perguntas e respostas

**1. O que é um Perceptron? (1,0 ponto)**

**Resposta:** O Perceptron é o modelo mais simples de rede neural artificial: um único neurônio que recebe entradas numéricas, multiplica cada uma por um peso, soma tudo (mais o viés, quando existe) e passa o resultado por uma função de ativação para produzir uma saída. É um classificador binário linear, que aprende ajustando seus pesos a partir dos erros cometidos.

**2. Quais são os principais elementos que compõem um Perceptron? (1,0 ponto)**

**Resposta:** Entradas (x1, x2, ...), pesos (w1, w2, ...), soma ponderada (entradas × pesos), viés/bias (termo constante somado), função de ativação (no nosso caso o degrau) e a saída (previsão). Para o treino também entram a taxa de aprendizado e a regra de atualização dos pesos.

**3. Qual é a função dos pesos dentro do Perceptron? (1,0 ponto)**

**Resposta:** Os pesos indicam a importância de cada entrada para a decisão. Um peso maior aumenta a influência daquela entrada na soma; um peso negativo a inibe. São os pesos que o Perceptron ajusta durante o treinamento para reduzir os erros, e é neles que fica o "conhecimento" aprendido.

**4. Qual é a função da função de ativação? (1,0 ponto)**

**Resposta:** Transformar a soma ponderada em uma saída final. A função degrau decide: se a soma é >= 0 a saída é 1 (liberar oferta), caso contrário é 0 (não liberar). Ela introduz o limiar de decisão que permite ao Perceptron classificar.

**5. Quantas épocas foram necessárias para o Perceptron convergir sem viés? (0,8 ponto)**

**Resposta:** Nenhuma: o Perceptron sem viés não convergiu. Nas 10 épocas executadas, o resultado foi sempre o mesmo (4 erros por época), com pesos voltando a [0.0, 0.0] ao final de cada época.

**6. O modelo conseguiu chegar a uma época com zero erros sem viés? (0,6 ponto)**

**Resposta:** Não. A quantidade de erros ficou em 4 em todas as 10 épocas, sem nunca chegar a zero.

**7. Explique por que esse comportamento ocorreu. Observe principalmente a entrada [0, 0]. (0,6 ponto)**

**Resposta:** Sem viés, a soma é w1·x1 + w2·x2. Para a entrada [0, 0] a soma é sempre 0, independentemente dos pesos, e degrau(0) = 1, mas o esperado é 0. Além disso, como a entrada é zero, a atualização (peso + taxa × erro × entrada) não altera nenhum peso, então o erro nunca é corrigido. Esse erro fixo impede qualquer época com zero erros. Os demais registros ainda ficam oscilando e os pesos voltam a [0, 0] a cada época. Faltava um termo independente das entradas para deslocar o limiar de decisão.

**8. Quantas épocas foram necessárias para o modelo convergir com viés? (0,6 ponto)**

**Resposta:** 4 épocas. Os erros por época foram 2, 3, 3 e 0; a época 4 foi a primeira sem erros.

**9. Quais foram os pesos finais encontrados? (0,4 ponto)**

**Resposta:** w1 = 0.2 e w2 = 0.1, ou seja, pesos finais [0.2, 0.1].

**10. Qual foi o valor final do bias? (0,4 ponto)**

**Resposta:** bias = -0.2.

**11. Para que serve o viés em um Perceptron? Explique com suas palavras. (0,6 ponto)**

**Resposta:** O viés é um valor somado à soma ponderada que desloca o limiar de ativação, independentemente das entradas. Ele permite que o neurônio responda 0 mesmo quando todas as entradas são 0 (como em [0, 0]) e dá liberdade para a reta de decisão não ser obrigada a passar pela origem. No AND, o viés negativo (-0.2) faz com que só a combinação [1, 1] consiga superar o limiar.

**12. Qual foi a principal diferença entre o treinamento sem viés e o treinamento com viés? (0,6 ponto)**

**Resposta:** Sem viés o modelo ficou preso em 4 erros por época durante as 10 épocas e não aprendeu o AND. Com viés ele convergiu em 4 épocas, com zero erros, classificando corretamente os 4 registros. A diferença é o termo extra (bias), que também é atualizado a cada erro (bias = bias + taxa × erro).

**13. O viés substitui os pesos? Explique. (0,4 ponto)**

**Resposta:** Não. Os dois têm papéis complementares: os pesos controlam a influência de cada entrada, enquanto o viés desloca o limiar de decisão e não depende de nenhuma entrada. O modelo precisa dos dois; no resultado final, w = [0.2, 0.1] e bias = -0.2 trabalham juntos.

**14. Um único Perceptron com viés seria suficiente para resolver um problema XOR? Explique. (0,4 ponto)**

**Resposta:** Não. O XOR não é linearmente separável: não existe uma única reta que separe as saídas 1 ([0,1] e [1,0]) das saídas 0 ([0,0] e [1,1]). Um Perceptron simples, mesmo com viés, só traça uma reta de decisão, portanto nunca chegaria a zero erros. Seria necessária uma rede com camada oculta (Perceptron multicamadas).

##Conclusão
Os pesos definem quanto cada entrada influencia a decisão, e o erro (esperado − previsto) indica o quanto e em que direção ajustá-los; cada passagem completa pelos dados é uma época. O treino converge quando uma época termina com zero erros, o que aconteceu apenas no modelo com viés, em 4 épocas (pesos [0.2, 0.1], bias −0.2). Sem viés, a entrada [0, 0] sempre produz soma 0, ativa o degrau em 1 e nunca é corrigida, então o modelo travou em 4 erros por época. Aprendi que o viés não substitui os pesos, ele os complementa deslocando o limiar de decisão, e que o Perceptron só resolve problemas linearmente separáveis, como o AND, mas não o XOR.
